```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('well_15_9-19A.csv')

# Part 1: Display all logs side by side

logs = ['GR','CALI','RHOB','NPHI','RT']

fig_logs,axes_logs = plt.subplots(1,5)
for i in range (5):
 axes_logs[i].plot(df[logs[i]],df['DEPTH'])
 axes_logs[i].invert_yaxis()
 axes_logs[i].set_xlabel(f'{logs[i]} LOG')

axes_logs[0].set_ylabel('DEPTH')
plt.tight_layout()

# Part 2: GR distribution

fig_hist , axes_hist = plt.subplots(1 , 1)
axes_hist.hist(df['GR'],bins=5)
axes_hist.set_xlabel('GR (API)')
axes_hist.set_ylabel('Frequency')

plt.tight_layout()

# Part 3: RHOB-NPHI crossplot

fig_scatter , axes_scatter = plt.subplots(1,1)
axes_scatter.scatter(df['NPHI'],df['RHOB'])
axes_scatter.set_ylabel('NPHI')
axes_scatter.set_xlabel('RHOB')
axes_scatter.set_title('Neutron-Density LOG')





plt.tight_layout()
plt.show()
```
